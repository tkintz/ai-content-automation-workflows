```python
from pathlib import Path
import os
PROJECT_DIR = Path(os.environ.get('SJK_PORTFOLIO_DIR', '/content/sjk_portfolio')) / 'Transcribe_mp4'
PROJECT_DIR.mkdir(parents=True, exist_ok=True)
import os
import whisper

# Initialize model
model = whisper.load_model("medium")

# Specify the directories
input_directory = str(PROJECT_DIR / 'mp4' / 'new')
output_directory = str(PROJECT_DIR / 'text')

# Iterate over all files in the input_directory
os.makedirs(input_directory, exist_ok=True)
os.makedirs(output_directory, exist_ok=True)
for filename in os.listdir(input_directory):
    # Check if file is a .mp3 file
    if filename.endswith(".mp4"):
        # Construct full file path
        file_path = os.path.join(input_directory, filename)

        # Transcribe the file
        result = model.transcribe(file_path)

        # Retrieve the filename without extension
        base_name = os.path.splitext(filename)[0]

        # Append .txt extension to the filename and join it with directory
        output_filename = os.path.join(output_directory, base_name + ".txt")

        # Write the transcription to the .txt file
        with open(output_filename, 'w') as output_file:
            output_file.write(result["text"])
```
