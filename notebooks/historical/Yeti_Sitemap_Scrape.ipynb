```python
from pathlib import Path
import os

# Portable output workspace; override SJK_PORTFOLIO_DIR to use a different folder.
PROJECT_DIR = Path(os.environ.get('SJK_PORTFOLIO_DIR', '/content/sjk_portfolio')) / 'Yeti_Sitemap_Scrape'
PROJECT_DIR.mkdir(parents=True, exist_ok=True)
path = str(PROJECT_DIR)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Read in the CSV file containing URLs
sitemap_urls = pd.read_csv(str(PROJECT_DIR / 'TA_sitemap_urls.csv'), header=None, names=['url'])

# Create an empty DataFrame to store the scraped data
scraped_data = pd.DataFrame(columns=['title', 'href', 'h2', 'article'])

# Loop through each URL in the CSV file
for index, row in sitemap_urls.iterrows():
    url = row['url']

    # Send a request to the URL and parse the HTML with BeautifulSoup
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')

    # Scrape the title and store it in the DataFrame along with the URL it was read from
    title = soup.find('h1').text.strip()
    href = url

    # Scrape the text from h2 tags
    h2_tags = soup.find_all('h2')
    for h2_tag in h2_tags:
        h2 = h2_tag.text.strip()
        scraped_data = scraped_data.append({'title': title, 'href': href, 'h2': h2, 'article': ''}, ignore_index=True)

    # Scrape the text between header tags and store them in the article column
    article = ''
    for header in soup.find_all(['h1', 'h2']):
        for elem in header.next_siblings:
            if elem.name in ['h1', 'h2']:
                break
            elif elem.string:
                article += elem.string.strip() + ' '
    # Append the article text to the existing row if the title and href match
    if not scraped_data.empty and scraped_data['title'].iloc[-1] == title and scraped_data['href'].iloc[-1] == href:
        scraped_data['article'].iloc[-1] += '\n' + article.strip()
    # Add a new row if the title or href are different
    else:
        scraped_data = scraped_data.append({'title': title, 'href': href, 'h2': '', 'article': article.strip()}, ignore_index=True)

# Save the scraped data to a new CSV file
scraped_data.to_csv(str(PROJECT_DIR / 'TA_sitemap_urls_test.csv'), index=False)
```
