```python
from pathlib import Path
import os

# Portable output workspace; override SJK_PORTFOLIO_DIR to use a different folder.
PROJECT_DIR = Path(os.environ.get('SJK_PORTFOLIO_DIR', '/content/sjk_portfolio')) / 'Yeti_Video_urls'
PROJECT_DIR.mkdir(parents=True, exist_ok=True)
path = str(PROJECT_DIR)
```

```python
import csv

input_file = str(PROJECT_DIR / 'SMED Videos and Links.txt')
output_file = str(PROJECT_DIR / 'Yet_Video_urls.csv')

# Read URLs from the input text file
with open(input_file, 'r') as file:
    urls = file.readlines()

# Remove newline characters from URLs
urls = [url.strip() for url in urls]

# Write URLs to the output CSV file
with open(output_file, 'w', newline='') as file:
    writer = csv.writer(file)
    for url in urls:
        writer.writerow([url])
```
